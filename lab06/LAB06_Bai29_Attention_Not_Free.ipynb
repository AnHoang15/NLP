```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

# Bai 29: so phan tu trong attention matrix tang theo T^2
sequence_lengths = [1_000, 10_000]
matrix_elements = [T ** 2 for T in sequence_lengths]
results = pd.DataFrame({
    "Sequence length (T)": [f"{T:,}" for T in sequence_lengths],
    "Attention matrix": [f"{T:,} x {T:,}" for T in sequence_lengths],
    "Elements (T^2)": [f"{n:,}" for n in matrix_elements],
    "FP32 memory (decimal MB)": [f"{n * 4 / 1_000_000:,.0f}" for n in matrix_elements],
})

display(Markdown("## Bai 29 - Attention khong mien phi"))
display(results)
growth = matrix_elements[1] / matrix_elements[0]
print(f"T tang tu {sequence_lengths[0]:,} len {sequence_lengths[1]:,} (gap {sequence_lengths[1] / sequence_lengths[0]:.0f})")
print(f"So phan tu ma tran tang gap {growth:.0f} lan, theo quy luat T^2.")
print("Bo nho FP32 ben tren chi tinh mot ma tran cho mot mau va mot head.")

plt.figure(figsize=(6, 3.5))
plt.loglog(sequence_lengths, matrix_elements, marker="o", color="#2858B8")
plt.xticks(sequence_lengths, [f"{T:,}" for T in sequence_lengths])
plt.yticks(matrix_elements, [f"{n:,}" for n in matrix_elements])

plt.xlabel("Sequence length (T)")
plt.ylabel("Attention matrix elements (T^2)")
plt.title("Quadratic growth of full self-attention")
plt.grid(True, which="both", linestyle=":", alpha=0.5)
plt.tight_layout()
plt.show()
```
