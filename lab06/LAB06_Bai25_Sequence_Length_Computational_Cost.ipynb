```python
# Bài 25 — Sequence length và computational cost
# Số phần tử attention matrix tăng theo T^2.

import pandas as pd

sequence_lengths = [10, 100, 500, 1000]
rows = []
for length in sequence_lengths:
    rows.append({
        'Sequence length (T)': length,
        'Attention matrix': f'{length} x {length}',
        'Number of elements (T^2)': length ** 2,
    })

cost_table = pd.DataFrame(rows)
display(cost_table)

print('Growth compared with T=10:')
for length in sequence_lengths:
    relative_growth = (length ** 2) / (sequence_lengths[0] ** 2)
    print(f'T={length:>4}: {length ** 2:>10,} elements ({relative_growth:,.0f}x)')

print('''
Câu hỏi cuối bài:
Self-attention phải tính tương tác giữa từng cặp token, nên attention matrix có T x T phần tử và chi phí bộ nhớ/tính toán tăng theo bình phương độ dài chuỗi, O(T^2). Khi chuỗi dài gấp 10 lần, số phần tử ma trận tăng gấp 100 lần. Vì vậy bộ nhớ và thời gian xử lý trở thành hạn chế đáng kể, đặc biệt với batch lớn và nhiều attention head.
''')
```
