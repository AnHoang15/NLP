# 23 Experiment 6 — Error Analysis


Các chỉ số và bảng bên dưới được tính trực tiếp từ nhãn và dự đoán đã lưu. Cột giải thích để trống cho sinh viên tự hoàn thiện.


```python
from google.colab import drive
drive.mount('/content/drive')
DRIVE_DIR = '/content/drive/MyDrive/NLP/lab09'
import os, json, random, numpy as np
os.makedirs(DRIVE_DIR, exist_ok=True)
SEED = 42
random.seed(SEED); np.random.seed(SEED)
import pandas as pd, json, numpy as np
def spans(words, tags):
    out=[]; start=None; typ=None
    def close(end):
        if start is not None: out.append((start,end,typ,' '.join(words[start:end])))
    for i, tag in enumerate(list(tags)+['O']):
        if tag == 'O':
            close(i); start=None; typ=None; continue
        prefix, kind = tag.split('-',1)
        if prefix == 'B' or typ != kind:
            close(i); start=i; typ=kind
        elif start is None:
            start=i; typ=kind
    return out

def entity_set(words, tags):
    return {(a,b,t) for a,b,t,_ in spans(words,tags)}

def prf(tp, fp, fn):
    p=tp/(tp+fp) if tp+fp else 0.0
    r=tp/(tp+fn) if tp+fn else 0.0
    return p,r,(2*p*r/(p+r) if p+r else 0.0)

def extract(words, tags):
    mapping={'PER':'PERSON','ORG':'ORG','LOC':'LOCATION','MISC':'MISC'}
    result={'PERSON':[],'ORG':[],'LOCATION':[],'MISC':[]}
    for a,b,typ,text in spans(words,tags):
        key=mapping[typ]
        if text not in result[key]: result[key].append(text)
    return result

def counts(gold_rows, pred_rows, type_filter=None):
    tp=fp=fn=0
    for g,p in zip(gold_rows,pred_rows):
        gs=entity_set(g['words'],g['gold']); ps=entity_set(p['words'],p['pred'])
        if type_filter:
            gs={x for x in gs if x[2]==type_filter}; ps={x for x in ps if x[2]==type_filter}
        tp+=len(gs&ps); fp+=len(ps-gs); fn+=len(gs-ps)
    return tp,fp,fn

with open(f'{DRIVE_DIR}/test_gold.json',encoding='utf-8') as f: gold=json.load(f)
with open(f'{DRIVE_DIR}/transformer_predictions.json',encoding='utf-8') as f: transformer=json.load(f)
assert len(gold)==len(transformer)
```

```python
items=[]
for sid,(g,p) in enumerate(zip(gold,transformer)):
    gs=spans(g['words'],g['gold']); ps=spans(p['words'],p['pred'])
    gset={(a,b,t) for a,b,t,_ in gs}; pset={(a,b,t) for a,b,t,_ in ps}
    for a,b,t,s in ps:
        if (a,b,t) not in gset: items.append({'sentence_id':sid,'context':' '.join(g['words']),'text':s,'gold':'','prediction':t,'error_type':'false positive','explanation':''})
    for a,b,t,s in gs:
        if (a,b,t) not in pset: items.append({'sentence_id':sid,'context':' '.join(g['words']),'text':s,'gold':t,'prediction':'','error_type':'false negative','explanation':''})
    for a,b,t,s in gs:
        for c,d,u,z in ps:
            if max(a,c)<min(b,d) and (a,b)!=(c,d): items.append({'sentence_id':sid,'context':' '.join(g['words']),'text':s,'gold':t,'prediction':f'{z} [{u}]','error_type':'boundary error','explanation':''})
            elif max(a,c)<min(b,d) and (a,b)==(c,d) and t!=u: items.append({'sentence_id':sid,'context':' '.join(g['words']),'text':s,'gold':t,'prediction':u,'error_type':'wrong type','explanation':''})
err=pd.DataFrame(items).drop_duplicates()
for kind in ['false positive','false negative','boundary error']:
    part=err[err.error_type==kind].head(5)
    print(kind,'available:',int((err.error_type==kind).sum()))
    display(part)
    assert len(part)>=5, f'Fewer than 5 {kind}s found'
err.to_csv(f'{DRIVE_DIR}/error_analysis.csv',index=False)
```
